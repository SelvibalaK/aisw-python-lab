```python
# string indexing and slicing
```

```python
text = "Python"
```

```python
# Indexing
print(text[0])     # P
print(text[-1])    # n
```

```python
# Slicing
print(text[0:4])   # Pyth
print(text[2:])    # thon
print(text[:3])    # Pyt
```

```python
# Reverse
print(text[::-1])  # nohtyP
```

```python
# string methods
```

```python
sentence = "        The quick brown fox jumped over the lazy dog      "
```

```python
# upper () - convert to uppercase

print(sentence.upper())
```

```python
# lower () - convert to lowercase

print(sentence.lower())
```

```python
# strip () - remove spaces that are in front and end

print(sentence.strip())
```

```python
# replace () - replaces the text

print(sentence.replace("jumped", "jumps"))
```

```python
# split () - converts to list 

print(sentence.split())
```

```python
# string examples 
```

```python
# example 1 - email validation checking 

email = " user@gmail.com "

if "@" in email and "." in email:
    print(" Valid email ")
else:
    print(" Invalid email ")
```

```python
# example 2 - username space stripping

username = "  Lee Dong Wook        "
clean_username = username.strip().lower()

print(clean_username)
```

```python
# example 3 - words counting 

text = " Almost nothing was more annoying than having our wasted time wasted on something not worth wasting it on "
words = text.lower().split()

print(words.count("wasted"))
```

```python
# example 4 - passwrod strength

password = "Python@123"

if len(password) >= 8:
    print("Strong password")
else:
    print("Weak password")
```
