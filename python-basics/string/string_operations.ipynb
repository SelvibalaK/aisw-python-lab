```python
# string indexing and slicing
```

```python
text = "Python"
```

```python
# Indexing
print(text[0])     # P
print(text[-1])    # n
```

```python
# Slicing
print(text[0:4])   # Pyth
print(text[2:])    # thon
print(text[:3])    # Pyt
```

```python
# Reverse
print(text[::-1])  # nohtyP
```
